# E-Commerce Purchase Conversion Analytics
### Data Analytics with AI Internship Capstone

This notebook analyzes online shopping sessions and converts behavioral patterns into business insights and recommended actions.

```python
!pip install -q ucimlrepo pandas numpy matplotlib plotly
from ucimlrepo import fetch_ucirepo
import pandas as pd, numpy as np
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
ds = fetch_ucirepo(id=468)
df = pd.concat([ds.data.features.copy(), ds.data.targets.copy()], axis=1)
df.head()
```

## 1. Data Quality and KPIs
The dataset contains 12,330 sessions. The target variable is `Revenue`, indicating whether the session ended in a purchase.

```python
df['Revenue'] = df['Revenue'].astype(str).str.lower().isin(['true','1','yes'])
df['Weekend'] = df['Weekend'].astype(str).str.lower().isin(['true','1','yes'])
print('Rows:', len(df))
print('Columns:', len(df.columns))
print('Missing values:', int(df.isna().sum().sum()))
print('Purchases:', int(df['Revenue'].sum()))
print('Conversion rate:', round(df['Revenue'].mean()*100,2), '%')
```

## 2. Conversion by Visitor Type
Compare new and returning visitors to identify segment-level opportunities.

```python
visitor = df.groupby('VisitorType')['Revenue'].mean().mul(100).reset_index(name='Conversion Rate')
px.bar(visitor, x='VisitorType', y='Conversion Rate', title='Conversion Rate by Visitor Type').show()
```

## 3. Monthly Conversion
Monthly patterns can support campaign scheduling and resource planning.

```python
monthly = df.groupby('Month')['Revenue'].mean().mul(100).reset_index(name='Conversion Rate')
px.bar(monthly, x='Month', y='Conversion Rate', title='Conversion Rate by Month').show()
```

## 4. Weekday vs Weekend
Compare conversion across visit timing.

```python
weekend = df.groupby('Weekend')['Revenue'].mean().mul(100).reset_index(name='Conversion Rate')
weekend['Visit Timing'] = weekend['Weekend'].map({True:'Weekend', False:'Weekday'})
px.bar(weekend, x='Visit Timing', y='Conversion Rate', title='Weekday vs Weekend Conversion').show()
```

## 5. Page Value Signals
Page-value bands provide a practical way to examine whether sessions showing stronger commercial intent also convert at higher rates.

```python
bins=[-0.001,0,5,20,50,np.inf]
labels=['0','0–5','5–20','20–50','50+']
df['Page Value Band']=pd.cut(df['PageValues'],bins=bins,labels=labels)
pv=df.groupby('Page Value Band',observed=False)['Revenue'].mean().mul(100).reset_index(name='Conversion Rate')
px.bar(pv,x='Page Value Band',y='Conversion Rate',title='Conversion Rate by Page Value Band').show()
```

## 6. Business Insights and Actions
- Track conversion rate as the primary outcome KPI.
- Compare new versus returning visitors to separate acquisition and retention opportunities.
- Use monthly conversion patterns to inform campaign timing.
- Use engagement and page-value signals to identify high-intent sessions.
- Combine the dashboard with operational actions such as targeted offers, landing-page optimization and conversion-focused UX improvements.
